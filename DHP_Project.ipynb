```python
import pandas as pd 
data = pd.read_csv("Telecome.csv")
print(data)
```

```python
#1.CALCULATE TOTAL REVENUE BY CONTRACT TYPE.
revenue = data.groupby('Contract')['TotalCharges'].sum()
print(revenue)
```

```python
#2.FIND AVERAGE MONTHLY CHARGES BY CONTRACT TYPE.
charges = data.groupby('Contract')['MonthlyCharges'].mean()
print(charges)
```

```python
#3.CALCULATE CHURN RATE.
total=len(data)
churn=len(data[data['Churn']=='Yes'])
churn_rate=(churn/total)*100
print(churn_rate)
```

```python
#4.IDENTIFY CUSTOMERS WITH TENURE GREATER THAN 24 MONTHS
customers = data[data['tenure'] > 24]
print(customers)
```

```python
#5.FIND TOP 10 HIGHEST PAYING CUSTOMERS (BASED ON TOTAL CHARGES).
sorted = data.sort_values('TotalCharges', ascending=False)
customers = sorted.head(10)
print(customers)
```

```python
#6.COMPARE AVERAGE MONTHLY CHARGES BETWEEN CHURN AND NON-CHURN CUSTOMERS.
avg_bill = data.groupby('Churn')['MonthlyCharges'].mean()
print(avg_bill)
```

```python
#7.FIND THE CONTRACT TYPE WITH THE HIGHEST CHURN RATE.
total_customer = data.groupby('Contract')['customerID'].count()
churned_customer = data[data['Churn'] == 'Yes'].groupby('Contract')['customerID'].count()
rate = (churned_customer / total_customer) * 100
sorted = rate.sort_values(ascending=False)
print(sorted.head(1))
```

```python
#8.IDENTIFY CUSTOMERS WITH MONTHLY CHARGES GREATER THAN 80.
customers = data[data['MonthlyCharges'] > 80]
print(customers)
```

```python
#9.CALCULATE AVERAGE TENURE BY CONTRACT TYPE.
avg_tenure = data.groupby('Contract')['tenure'].mean()
print(avg_tenure)
```

```python
#10.FIND TOTAL CUSTOMERS PER CONTRACT TYPE.
customers = data["Contract"].value_counts()
print(customers)
```
